```python
'''AI Tic-Tac-Toe Game
Play Tic-Tac-Toe against a simple computer opponent.

You play as X.
The computer plays as O.
Enter a position from 1 to 9.'''
```

```python

import math

HUMAN = "X"
AI = "O"
EMPTY = " "

def check_winner(board):
    wins = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    for a, b, c in wins:
        if board[a] == board[b] == board[c] != EMPTY:
            return board[a]

    return None


def is_draw(board):
    return EMPTY not in board and check_winner(board) is None


def minimax(board, depth, maximizing):
    winner = check_winner(board)

    if winner == AI:
        return 10 - depth
    if winner == HUMAN:
        return depth - 10
    if is_draw(board):
        return 0

    if maximizing:
        best_score = -math.inf

        for i in range(9):
            if board[i] == EMPTY:
                board[i] = AI
                score = minimax(board, depth + 1, False)
                board[i] = EMPTY
                best_score = max(best_score, score)

        return best_score

    else:
        best_score = math.inf

        for i in range(9):
            if board[i] == EMPTY:
                board[i] = HUMAN
                score = minimax(board, depth + 1, True)
                board[i] = EMPTY
                best_score = min(best_score, score)

        return best_score


def best_move(board):
    best_score = -math.inf
    move = None

    for i in range(9):
        if board[i] == EMPTY:
            board[i] = AI
            score = minimax(board, 0, False)
            board[i] = EMPTY

            if score > best_score:
                best_score = score
                move = i

    return move


print("AI algorithm loaded successfully!")
```

```python
import ipywidgets as widgets
from IPython.display import display

board = [EMPTY] * 9
game_over = False

status = widgets.HTML("<h3>Your turn! You are X.</h3>")

buttons = [
    widgets.Button(
        description=" ",
        layout=widgets.Layout(width="80px", height="65px")
    )
    for i in range(9)
]


def update_board():
    for i, button in enumerate(buttons):
        button.description = board[i] if board[i] != EMPTY else " "
        button.disabled = game_over or board[i] != EMPTY


def finish_game():
    global game_over

    winner = check_winner(board)

    if winner == HUMAN:
        status.value = "<h3>You win! Congratulations!</h3>"
        game_over = True
    elif winner == AI:
        status.value = "<h3>AI wins! Better luck next time!</h3>"
        game_over = True
    elif is_draw(board):
        status.value = "<h3>It's a draw!</h3>"
        game_over = True


def handle_click(index):
    global game_over

    if game_over or board[index] != EMPTY:
        return

    board[index] = HUMAN
    update_board()

    if check_winner(board) == HUMAN or is_draw(board):
        finish_game()
        update_board()
        return

    status.value = "<h3>AI is thinking...</h3>"

    move = best_move(board)

    if move is not None:
        board[move] = AI

    finish_game()

    if not game_over:
        status.value = "<h3>Your turn! You are X.</h3>"

    update_board()


for i, button in enumerate(buttons):
    button.on_click(lambda b, i=i: handle_click(i))


reset_button = widgets.Button(
    description="New Game",
    button_style="success"
)


def reset_game(b):
    global board, game_over

    board = [EMPTY] * 9
    game_over = False
    status.value = "<h3>Your turn! You are X.</h3>"
    update_board()


reset_button.on_click(reset_game)

game_grid = widgets.GridBox(
    children=buttons,
    layout=widgets.Layout(
        grid_template_columns="80px 80px 80px",
        grid_template_rows="65px 65px 65px",
        grid_gap="4px"
    )
)

game_ui = widgets.VBox([status, game_grid, reset_button])

update_board()

print("Game interface created successfully!")
```
